Arithmetic operator
+,-,*,/,//,**

```python
a=8
b=2
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a**b)
```

comparison operator
==,!=,>,<,<=,>=

```python
x=10
y=20
z=30
print(x==y)
print(x!=y)
print(z>y)
print(y>=x)
print(z>y>x)
print(x<y>z)
print(x<y<z)
print(x<=y)
```

assignment operator
=,+=,-=,*=,/=,%=,**=

```python
a=6
a+=3   #a=a+3
print(a)

a-=3
print(a)

a/=3
print(a)

a*=3
print(a)

a**=3
print(a)

a//=3
print(a)
```

identity operator  is,not is

```python
c=300
d=300
print(c is d)
print(c is not d)
print(id(c))
print(id(d))

a=100
b=a
print(a is b)
print(a is not b)
```

```python
a=[1,2,3]
b=a
c=[1,2,3]
print(a is b)
print(a is not b)
print(c is a)
print(c is not a)
```

membership operator
in ,not in

```python
a=[1,2,3]
print(1 in a)

l=["apple","orange"]
print("apple" in l)
print("apple" not in l)
print("banana" in l)


p="alida"
print("i" in p)

tu=(1,"lava","place")
print(2 in tu)
print("lava" in tu)
```

logical operator

```python
a=10
b=3
print(a>b and a<b)
print(b>a or b<a)
print(not a>b)
print(a>b and a==10)
```

bitwise operator
convert to bit then perform and,or..

```python
a=10
b=4
print(a&b)
print(a|b)
print(a^b)
print(~a)             #~a = -(a + 1)   ~10= -(10 +1)
print(~b)
print(a<<2)
print(a>>2)
```
