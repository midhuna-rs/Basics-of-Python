```python
s={1,4,1,5,6,7,2}
print(s)

#adding element
s.add(65)
print(s)

s.add((8,76))
print(s)

s.remove(1)       #error if we try to remove a element that is not present
print(s)
# s.remove(9)
# print(s)


s.discard(9)      #no error if we try to remove a element that is not present
print(s)
s.discard(65)
print(s)

#pop() will delete a random element
s.pop()
print(s)

s.clear()      #clear empties the set
print(s)
```

set operations

```python
a={1,2,3,1,5}
b={3,5,4,7,8}
print("union:",a | b)                      #all elements in both sets
print("intersection:",a & b)               #common elements in both sets
print("difference:",a - b)                 #in 'a' but not in 'b'
print("symmetric difference:",a ^ b)       #non common elements
```

```python
a={1,2,3,1,5}
b={3,5,4,7,8}
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
```

```python
a={1,2,3,1,5}
b={3,5,4,7,8}
c={1,4,7,9,5,8}
d=a|b|c
print(d)
e=a^b^c            #^ means symmetric difference: values occurring an odd number of times across the sets.
print(e)
```

```python
s1={"lily","rose","jasmine"}
s2={"lily","hibiscus","orchid"}
s3={"coco","hibiscus","lily"}
print(s1.union(s2,s3))
print(s1.intersection(s2,s3))
print(s1.difference(s2,s3))
print(s1.symmetric_difference(s2,s3))
```

## DICTIONARY
key value pairs,mutable,unordered,keys are unique and immutable

```python
d1={"name":"alice","age":20}
print(d1)
print(type(d1))

#adding key value
d1["place"]="karnataka"
print(d1)

#using dict()
d2=dict({"name":"banu","place":"kerala"})
print(d2)

d4=dict(a=1,b=2,c=3)
print(d4)

#empty dictionary
d3={}
print(d3)

```

```python
d1={"name":"alice","age":20,"marks":90}

#accessing elements
print(d1["name"])

#accessing a value using get method
print(d1.get("name"))
print(d1.get("city"))   #if not present = none

print(d1.get("name",0))
print(d1.get("city",0))   #if not present = 0

#value updation , adding
d1["name"]="giya"
d1["course"]="data science"
print(d1)

#deleting a (value)
d1.pop("name")
print(d1)

#to delete last inserted value
d1.popitem()
print(d1)

#del to delete a specific key value and entire dictionary
del d1["marks"]
print(d1)

#clear()
d1.clear()
print(d1)

#to delete entire dictionary
del d1
print(d1)
```

```python
d1={"name":"alice","age":20,"marks":90}
print(d1.keys())         #to print all keys
print(d1.values())       #to print all values
print(d1.items())        #to print as key value pair    #tuple format
```

```python
#update
d1.update({"name":"niya"})
print(d1)

#copy
d2=d1.copy()
print(d2)
```
