**string methods**

```python
a="TryUser"
print("upper:",a.upper())            # Converts all letters to uppercase
print("lower:",a.lower())            # Converts all letters to lowercase
b="pYthOn ProGraMming"
print(b.title())                     # Converts the first letter of each word to uppercase
print(b.capitalize())                # Converts only the first character to uppercase
print(b.swapcase())                  # Converts uppercase letters to lowercase and lowercase to uppercase
c="Straße"  
print(c.casefold())                  # Converts the string to lowercase for case-insensitive comparison
```

## index value finding

### find()
The find() method returns the index of the first occurrence of the specified substring. If the substring is not found, it returns -1.

```python

a="python programming"
print(a.find('pr'))
print(a.find("u"))  #letter that is not present  -1

```

### index()
The index() method returns the index of the first occurrence of the specified substring. If the substring is not found, it raises a ValueError.

```python
s="banana"
print(s.index('a'))
#print(s.index('u'))   #letter that is not present gives 'valueerror'
```

### rfind()
right find is used to find the position of the last occurrence of a character or substring.

```python
s="banana"
print(s.rfind("a"))   
# print(s.lfind('a'))   #attributeerror

#rfind() does not use negative indexing. The r means it searches from the right side,
# but the result is still returned using the normal positive index.
```

### count()
The count() method is used to count how many times a character or substring occurs in a string.

```python
s="banana"
print(s.count('a'))
```

### in (membership operator)
'in' is used to check whether a character or substring exists in a string.

```python
s="banana"

print("b" in s)      #present character

print("r" in s)      #absent character
```

## split()
to break a string into a list of smaller strings.

```python
text="  python  is  fun  "
print(text.split())       # Splits by whitespace and removes extra spaces
print(text.split(" "))    # Splits specifically at each space
print(text.split("o"))    # Splits wherever "o" occurs
print(text.split("n"))    # Splits wherever "n" occurs  
```

## join()
to combine elements of a list into one string, using a separator.

```python
text="  python  is  fun  "

words = text.split()              # Converts the string into a list of words

print("".join(words))             # Joins words without any separator
print("$".join(words))            # Joins words using "$" as the separator
print("-".join(words))            # Joins words using "-" as the separator
```

## partition()
that splits a string into 3 parts using a separator:
before separator → separator → after separator

```python
a = "hello-world-python"

print(a.partition("-"))   # Splits at the first "-" into: before, separator, after
print(a.partition("o"))   # Splits at the first "o" into: before, separator, after
```

## strip()
used to remove whitespace from both the beginning and the end of a string.

```python
a="  hello world  "

print(a.strip())    # Removes spaces from both sides
print(a.lstrip())   # Removes spaces from the left side
print(a.rstrip())   # Removes spaces from the right side
```

## zfill()
adds zeros to the left side of a string until it reaches the specified length.

```python
a="57"
print(a.zfill(4))     # Adds zeros to the left to make the string length 4

b="apple"
print(b.zfill(7))
```
